```python
# Label Incoding
```

```python
df.shape
```

```python
df['Transmission'] = df['Transmission'].map({'Manual' : 0, 'Automatic': 1})
df['Selling_type'] = df['Selling_type'].map({'Dealer' : 0, 'Individual': 1})
df.head()
```

```python
df.head(15)
```

```python
# Onehot Incoding
```

```python
df = pd.get_dummies(df, columns=['Fuel_Type'],drop_first=True)
```

```python
df.head()
```

```python
dummy_cols = ['Fuel_Type_Diesel', 'Fuel_Type_Petrol']
df[dummy_cols] = df[dummy_cols].astype(int)
```

```python
df.head(10)
```

```python
df['Car_Name'].value_counts().head(20)
```

```python
top_brands = df['Car_Name'].value_counts().nlargest(12).index
df['Car_Name'] = df['Car_Name'].where(df['Car_Name'].isin(top_brands),'other')
df_encoded = pd.get_dummies(df, columns=['Car_Name'], drop_first=True, dtype=int)
```

```python
df_encoded.head(10)
```

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
cols = ['Year','Present_Price', 'Driven_kms']

df_encoded[cols] = scalar.fit_transform(df_encoded[cols])
```

```python
df_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X = df_encoded.drop('Selling_Price', axis = 'columns')
y = df_encoded['Selling_Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 1)
```

```python
len(X_train)
len(y_train)
```

```python
len(X_test)
len(y_test)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error

y_true = y_test
y_predic = model.predict(X_test)

mae = mean_absolute_error(y_true, y_predic)
mse = mean_squared_error(y_true, y_predic)
rmse = root_mean_squared_error(y_true, y_predic)
print(f"MAE: {mae:.2f}, Mse: {mse:.2f}, RSME: {rmse:.2f}")
```
